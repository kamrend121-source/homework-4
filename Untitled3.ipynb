```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

# Load Iris dataset
iris = load_iris()
X = iris.data

# Calculate WCSS for k = 1 through 10
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.title("The elbow method")
plt.xlabel("Number of clusters")
plt.ylabel("WCSS")
plt.xticks(range(1, 11))
plt.show()
```

```python
import time
import pandas as pd

from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Load MNIST
mnist = fetch_openml('mnist_784', version=1, as_frame=False)

X = mnist.data.astype('float32')
y = mnist.target

print("Dataset shape:", X.shape)

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# -------------------------------------------------
# WITHOUT PCA
# -------------------------------------------------

# Standardize
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Logistic Regression
logisticRegr = LogisticRegression(
    solver='lbfgs',
    max_iter=100
)

start_time = time.time()

logisticRegr.fit(X_train_scaled, y_train)
y_pred = logisticRegr.predict(X_test_scaled)

end_time = time.time()

time_without_pca = end_time - start_time

accuracy_without_pca = accuracy_score(y_test, y_pred)

print("WITHOUT PCA")
print("Accuracy:", accuracy_without_pca)
print("Time:", time_without_pca, "seconds")


# -------------------------------------------------
# WITH PCA
# -------------------------------------------------

# PCA retaining 95% of variance
pca = PCA(n_components=0.95)

start_pca = time.time()

X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

pca_time = time.time() - start_pca

print("\nOriginal features:", X_train_scaled.shape[1])
print("PCA features:", X_train_pca.shape[1])
print("PCA time:", pca_time, "seconds")


# Logistic Regression after PCA
logisticRegr_pca = LogisticRegression(
    solver='lbfgs',
    max_iter=100
)

start_time = time.time()

logisticRegr_pca.fit(X_train_pca, y_train)
y_pred_pca = logisticRegr_pca.predict(X_test_pca)

end_time = time.time()

time_with_pca = end_time - start_time

accuracy_with_pca = accuracy_score(y_test, y_pred_pca)

print("\nWITH PCA")
print("Accuracy:", accuracy_with_pca)
print("Logistic Regression Time:", time_with_pca, "seconds")
```

```python
results = pd.DataFrame({
    "Method": ["Without PCA", "With PCA (95%)"],
    "Number of Features": [
        X_train_scaled.shape[1],
        X_train_pca.shape[1]
    ],
    "Time (seconds)": [
        time_without_pca,
        time_with_pca
    ],
    "Accuracy": [
        accuracy_without_pca,
        accuracy_with_pca
    ]
})

print(results)
```
