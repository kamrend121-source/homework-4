```python
# ---------------------------------------------------------
# 1. Loading the Dataset
# ---------------------------------------------------------
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
iris = load_iris()
X = iris.data
y = iris.target

print("Dataset Loaded Successfully")
print("Feature Names:", iris.feature_names)
print("Target Names:", iris.target_names)

# ---------------------------------------------------------
# 2. Checking the Sample and Target Sizes
# ---------------------------------------------------------
print("\nSample Size (X):", X.shape)
print("Target Size (y):", y.shape)

# ---------------------------------------------------------
# 3. Splitting the Data for Training and Testing
# ---------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# ---------------------------------------------------------
# 4. Training and Testing Set Sizes
# ---------------------------------------------------------
print("\nTraining Set Size:", X_train.shape)
print("Testing Set Size:", X_test.shape)

# ---------------------------------------------------------
# 5. Creating the Model (GaussianNB)
# ---------------------------------------------------------
model = GaussianNB()

# ---------------------------------------------------------
# 6. Training the Model
# ---------------------------------------------------------
model.fit(X_train, y_train)
print("\nModel Training Complete")

# ---------------------------------------------------------
# 7. Predicting
# ---------------------------------------------------------
y_pred = model.predict(X_test)
print("\nPredictions:", y_pred)

# ---------------------------------------------------------
# 8. Estimator Method score
# ---------------------------------------------------------
score = model.score(X_test, y_test)
print("\nModel Accuracy (score method):", score)

# ---------------------------------------------------------
# 9. Confusion Matrix
# ---------------------------------------------------------
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:\n", cm)

# ---------------------------------------------------------
# 10. Classification Report
# ---------------------------------------------------------
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

# ---------------------------------------------------------
# 11. Visualizing the Confusion Matrix
# ---------------------------------------------------------
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, cmap="Blues", fmt="d",
            xticklabels=iris.target_names,
            yticklabels=iris.target_names)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# ---------------------------------------------------------
# 12. K-Fold Cross-Validation
# ---------------------------------------------------------
kfold = KFold(n_splits=10, shuffle=True, random_state=42)
cv_scores = cross_val_score(model, X, y, cv=kfold)

print("\nK-Fold Cross-Validation Scores:", cv_scores)
print("Average CV Score:", cv_scores.mean())
```
